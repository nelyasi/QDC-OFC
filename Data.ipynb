```python
# --- Style ---
plt.rcParams["font.family"] = "serif"
plt.rcParams["font.size"] = 16

# --- EPR pair formulas ---
def E_star_remote(n):
    return n - 1

def E_line_end_remote(n):
    return n * (n - 1) / 2

def E_ring_remote(n):
    # For a ring, even and odd n have slightly different closed forms
    if n % 2 == 0:
        return (n ** 2) / 4
    else:
        return ((n - 1) * (n + 1)) / 4

# --- Compute values ---
ns = np.arange(2, 8)
E_line = [E_line_end_remote(n) for n in ns]
E_ring = [E_ring_remote(n) for n in ns]
E_star = [E_star_remote(n) for n in ns]

# --- Plot ---
plt.figure(figsize=(6, 5))
plt.plot(ns, E_line, '-o', label='Line')
plt.plot(ns, E_ring, '-s', label='Ring')
plt.plot(ns, E_star, '-^', label='Star')

# --- Discrete axis ticks ---
plt.xticks(np.arange(1, 8, 1))                      # show 1, 2, 3, 4, 5, 6, 7
plt.yticks(np.arange(0, int(max(E_line)) + 2, 2))   # jump 2 by 2 on y-axis

plt.xlabel('Number of QPUs (n)')
plt.ylabel('Entanglement Links')
#plt.title('Total Commuincation Cost', fontsize = 16)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
